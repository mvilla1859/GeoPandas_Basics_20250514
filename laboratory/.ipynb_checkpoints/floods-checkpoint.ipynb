```python
import geopandas as gpd
import matplotlib.pyplot as plt  # only for colormap names, not strictly needed

infra = gpd.read_file("/Users/manuelviro/Downloads/Peak_Control_Infrastructure_2534346855449688894.geojson")
# Pick a qualitative palette with at least 3 distinct colors
# You can use 'Set2', 'Set1', 'tab10', etc. (matplotlib colormaps)
palette = "tab10"

# Build the interactive map:
m = infra.explore(
    # titles=None,
    # titles='Stamen Terrain',
    # titles='Stamen Watercolor',
    # titles='Stamen Toner',
    column="INFRASTRUC",     # color by this column
    titles='CartoDB positron',
    categorical=True,        # treat it as categories (not numeric)
    cmap=palette,            # qualitative colormap
    legend=True,             # show legend
    legend_kwds={"title": "Infrastructure"},
    tooltip=["INFRASTRUC"],  # show type on hover
    style_kwds={             # polygon/line styling (ignored for point markers)
        "fillOpacity": 0.6,
        "weight": 0.7,
        "opacity": 0.8
    },
    # If your geometries are points, uncomment and tweak marker settings:
    # marker_type="circle",
    # marker_kwds={"radius": 6}
)

m.save('seattle_drainage_capacity_constrained_areas.html')
m
```

```python
import geopandas as gpd
# 1. Make sure we're in WGS84 (lat/lon) so web tiles behave nicely
if infra.crs is not None and infra.crs.to_epsg() != 4326:
    infra = infra.to_crs(epsg=4326)
# 2. Define a stable colour mapping for each INFRASTRUC category
infra_color_map = {
    "Ditch": "brown",
    "Densmore": "orange",
    "Pike/Melrose": "purple",
    "South Park": "green",
}
# 3. Build a list of categories in the order we want the legend
infra_categories = list(infra_color_map.keys())
infra_colors = list(infra_color_map.values())
# 4. Create interactive map with GeoPandas .explore()
m = infra.explore(
    column="INFRASTRUC",             # colour features by this column
    categorical=True,                # treat it as categories, not a numeric scale
    categories=infra_categories,     # order of categories in legend
    cmap=infra_colors,               # colours in same order as categories
    tooltip=["INFRASTRUC"],          # show this field on hover
    popup=True,                      # full attribute popup on click
    tiles="CartoDB positron",        # nicer basemap than default OSM
    marker_type="marker",            # draw as marker points (if your geometry is points)
)
# 5. Save as self-contained web map
m.save("seattle_drainage_capacity_constrained_areas.html")
m
```

```python
import folium

# 1. Make sure infra is in WGS84 (lat/lon) so Folium (Leaflet) is happy
if infra.crs is not None and infra.crs.to_epsg() != 4326:
    infra = infra.to_crs(epsg=4326)

# 2. Compute a global centroid for the whole layer to center the map
layer_centroid = infra.geometry.unary_union.centroid
m = folium.Map(
    location=[layer_centroid.y, layer_centroid.x],
    zoom_start=12,
    tiles="CartoDB positron"  # nicer light basemap
)

# 3. Define a colour for each INFRASTRUC category
infra_color_map = {
    "Ditch": "brown",
    "Densmore": "green",
    "Pike/Melrose": "purple",
    "South Park": "orange",
}

# 4. Add each polygon as both:
#    - an outline/fill polygon
#    - a centroid marker to label / highlight the category
for _, row in infra.iterrows():
    geom = row.geometry
    if geom.is_empty:
        continue

    infra_type = row["INFRASTRUC"]
    color = infra_color_map.get(infra_type, "gray")

    # ---- 4a. Add the polygon itself ----
    # folium can render polygon geometries directly using GeoJson
    folium.GeoJson(
        geom.__geo_interface__,
        style_function=lambda feature, col=color: {
            "color": col,          # outline colour
            "weight": 1,           # outline width
            "fill": True,
            "fillColor": col,
            "fillOpacity": 0.4,    # semi-transparent fill
        },
        tooltip=f"{infra_type}",
        popup=folium.Popup(
            f"INFRASTRUC: {infra_type}",
            max_width=250
        ),
    ).add_to(m)

    # ---- 4b. Add a centroid marker for clarity (optional) ----
    # For polygons and multipolygons, use the centroid point
    # centroid = geom.centroid

    # folium.CircleMarker(
    #     location=[centroid.y, centroid.x],  # centroid has .x and .y
    #     radius=4,
    #     color=color,
    #     fill=True,
    #     fill_color=color,
    #     fill_opacity=0.9,
    #     tooltip=infra_type,
    # ).add_to(m)

# 5. Save to HTML
m.save("infra_map_folium.html")
m
```
